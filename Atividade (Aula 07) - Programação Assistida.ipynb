Soma Lista:

```python
# Definição de uma lista de números para exemplo
numeros = [10, 20, 30, 40, 50]

# --- Sua abordagem original (refatorada para funcionar) ---
soma_while = 0
i = 0
while i < len(numeros):
  soma_while += numeros[i]
  i += 1
print(f"Soma usando o loop while: {soma_while}")

# --- Abordagem Pythonica com a função sum() ---
soma_builtin = sum(numeros)
print(f"Soma usando a função sum(): {soma_builtin}")

# Explicação:
# 1. 'numeros = [10, 20, 30, 40, 50]': Define uma lista de exemplo para que o código possa ser executado.
# 2. 'soma_while = 0' e 'i = 0': Inicializam a variável que armazenará a soma e o índice para o loop.
# 3. 'while i < len(numeros)': O loop continua enquanto o índice 'i' for menor que o número total de elementos na lista.
# 4. 'soma_while += numeros[i]': Adiciona o elemento atual da lista (na posição 'i') à variável 'soma_while'.
# 5. 'i += 1': Incrementa o índice para passar para o próximo elemento da lista.
# 6. 'soma_builtin = sum(numeros)': Esta é a forma mais simples e recomendada em Python para somar todos os elementos de uma lista. A função `sum()` faz o trabalho pesado de forma eficiente.
```

Eh Primo:

```python
def eh_primo(n):
  # Um número primo é um número natural maior que 1 que não possui divisores positivos além de 1 e de si mesmo.

  # Caso 1: Números menores ou iguais a 1 não são primos.
  if n <= 1:
    return False

  # Caso 2: O número 2 é o único número primo par.
  if n == 2:
    return True

  # Caso 3: Qualquer número par maior que 2 não é primo.
  if n % 2 == 0:
    return False

  # Caso 4: Verifica divisores ímpares a partir de 3 até a raiz quadrada de n.
  # Não precisamos verificar números pares, pois já os tratamos no Caso 3.
  # Se n não for divisível por nenhum número ímpar nesse intervalo, ele é primo.
  # O loop vai até int(n**0.5) + 1 porque se um número n tem um divisor maior que sua raiz quadrada,
  # ele também deve ter um divisor menor que sua raiz quadrada.
  for i in range(3, int(n**0.5) + 1, 2):
    if n % i == 0:
      return False

  # Se o loop terminar sem encontrar nenhum divisor, o número é primo.
  return True
```

Inversão String:

```python
def inverte_texto(texto):
  # Inicializa uma string vazia para armazenar o texto invertido
  invertido = ''
  # Itera sobre cada letra na string original
  for letra in texto:
    # Adiciona a letra atual ao *início* da string 'invertido'
    # Isso efetivamente constrói a string de trás para frente
    invertido = letra + invertido
  # Retorna a string completamente invertida após o loop ter processado todas as letras
  return invertido
```

Calculadora Simples:

```python
# Solicita ao usuário o primeiro número e converte para float para permitir decimais.
num1 = float(input("Informe o primeiro número: ")) # Alterado de 'um número' para 'o primeiro número'

# Solicita ao usuário o segundo número e converte para float.
num2 = float(input("Informe o segundo número: ")) # Alterado de 'um número' para 'o segundo número'

# Solicita ao usuário a operação desejada.
operação = input("Informe uma operação matemática (+, -, *, //): ")

# Verifica qual operação foi informada e executa o cálculo correspondente.
# A comparação 'operação == "+"' garante que a string informada seja exatamente a esperada.
if operação == "+":
  print(f"Resultado: {num1 + num2}") # Uso de f-string para formatação clara do resultado
elif operação == "-":
  print(f"Resultado: {num1 - num2}") # Uso de f-string
elif operação == "*":
  print(f"Resultado: {num1 * num2}") # Uso de f-string
elif operação == "//":
  # Melhoria: Adiciona uma verificação para evitar a divisão por zero.
  # Dividir por zero causaria um 'ZeroDivisionError'.
  if num2 == 0:
    print("Erro: Não é possível dividir por zero!") # Mensagem de erro específica para divisão por zero
  else:
    print(f"Resultado: {num1 // num2}") # Uso de f-string e divisão inteira
else:
  # Caso nenhuma operação válida seja informada, exibe uma mensagem de erro mais descritiva.
  print("O operador informado é inválido. Por favor, use +, -, *, ou //.") # Mensagem de erro melhorada
```
