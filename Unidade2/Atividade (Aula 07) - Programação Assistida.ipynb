Soma Lista:

```python
# Definição de uma lista de números para exemplo
numeros = [10, 20, 30, 40, 50]

# --- Sua abordagem original (refatorada para funcionar) ---
soma_while = 0
i = 0
while i < len(numeros):
  soma_while += numeros[i]
  i += 1
print(f"Soma usando o loop while: {soma_while}")

# --- Abordagem Pythonica com a função sum() ---
soma_builtin = sum(numeros)
print(f"Soma usando a função sum(): {soma_builtin}")

# Explicação:
# 1. 'numeros = [10, 20, 30, 40, 50]': Define uma lista de exemplo para que o código possa ser executado.
# 2. 'soma_while = 0' e 'i = 0': Inicializam a variável que armazenará a soma e o índice para o loop.
# 3. 'while i < len(numeros)': O loop continua enquanto o índice 'i' for menor que o número total de elementos na lista.
# 4. 'soma_while += numeros[i]': Adiciona o elemento atual da lista (na posição 'i') à variável 'soma_while'.
# 5. 'i += 1': Incrementa o índice para passar para o próximo elemento da lista.
# 6. 'soma_builtin = sum(numeros)': Esta é a forma mais simples e recomendada em Python para somar todos os elementos de uma lista. A função `sum()` faz o trabalho pesado de forma eficiente.
```

Eh Primo:

```python
n = float(input("Informe um número: ")) # Solicita que o usuário informe o número que será analisado no loop.
def eh_primo(n):
  # Um número primo é um número natural maior que 1 que não possui divisores positivos além de 1 e de si mesmo.

  # Caso 1: Números menores ou iguais a 1 não são primos.
  if n <= 1:
    return False

  # Caso 2: O número 2 é o único número primo par.
  if n == 2:
    return True

  # Caso 3: Qualquer número par maior que 2 não é primo.
  if n % 2 == 0:
    return False

  # Caso 4: Verifica divisores ímpares a partir de 3 até a raiz quadrada de n.
  # Não precisamos verificar números pares, pois já os tratamos no Caso 3.
  # Se n não for divisível por nenhum número ímpar nesse intervalo, ele é primo.
  # O loop vai até int(n**0.5) + 1 porque se um número n tem um divisor maior que sua raiz quadrada,
  # ele também deve ter um divisor menor que sua raiz quadrada.
  for i in range(3, int(n**0.5) + 1, 2):
    if n % i == 0:
      return False

  # Se o loop terminar sem encontrar nenhum divisor, o número é primo.
  return True
print(eh_primo(n)) # Imprime o resultado com resposta final de True ou False.
```

Inversão String:

```python
texto = input("Informe uma frase: ") # Solicita que o usuário informe a frase que será invertida
def inverte_texto(texto):
  # Inicializa uma string vazia para armazenar o texto invertido
  invertido = ''
  # Itera sobre cada letra na string original
  for letra in texto:
    # Adiciona a letra atual ao *início* da string 'invertido'
    # Isso efetivamente constrói a string de trás para frente
    invertido = letra + invertido
  # Retorna a string completamente invertida após o loop ter processado todas as letras
  return invertido
print(inverte_texto(texto)) # Imprime a frase informada invertida
```

Calculadora Simples:

```python
# Solicita ao usuário o primeiro número e converte para float para permitir decimais.
num1 = float(input("Informe o primeiro número: ")) # Alterado de 'um número' para 'o primeiro número'

# Solicita ao usuário o segundo número e converte para float.
num2 = float(input("Informe o segundo número: ")) # Alterado de 'um número' para 'o segundo número'

# Solicita ao usuário a operação desejada.
operação = input("Informe uma operação matemática (+, -, *, //): ")

# Verifica qual operação foi informada e executa o cálculo correspondente.
# A comparação 'operação == "+"' garante que a string informada seja exatamente a esperada.
if operação == "+":
  print(f"Resultado: {num1 + num2}") # Uso de f-string para formatação clara do resultado
elif operação == "-":
  print(f"Resultado: {num1 - num2}") # Uso de f-string
elif operação == "*":
  print(f"Resultado: {num1 * num2}") # Uso de f-string
elif operação == "//":
  # Melhoria: Adiciona uma verificação para evitar a divisão por zero.
  # Dividir por zero causaria um 'ZeroDivisionError'.
  if num2 == 0:
    print("Erro: Não é possível dividir por zero!") # Mensagem de erro específica para divisão por zero
  else:
    print(f"Resultado: {num1 // num2}") # Uso de f-string e divisão inteira
else:
  # Caso nenhuma operação válida seja informada, exibe uma mensagem de erro mais descritiva.
  print("O operador informado é inválido. Por favor, use +, -, *, ou //.") # Mensagem de erro melhorada
```

Jogo da Forma:

```python
import random # Importa o módulo 'random' para usar funções de escolha e amostragem aleatórias.

formas = { # Define um dicionário chamado 'formas' onde as chaves são nomes de formas geométricas e os valores são listas de dicas sobre essas formas.
    "quadrado": ["4 lados iguais", "ângulos de 90 graus"],
    "triângulo": ["3 lados", "3 vértices", "soma dos ângulos internos é 180 graus"],
    "círculo": ["sem lados", "redondo", "raio"],
    "retângulo": ["4 lados", "lados opostos iguais", "4 ângulos de 90 graus"],
    "pentágono": ["5 lados", "5 vértices"]
}

pontuacao = 0 # Inicializa a pontuação do jogador com zero.
jogando = True # Define uma variável booleana para controlar o loop principal do jogo, iniciando como True para o jogo começar.

print("Bem-vindo ao Jogo da Forma! Tente identificar a forma geométrica pelas dicas.") # Exibe uma mensagem de boas-vindas e instruções iniciais.
print("Para sair, digite 'sair'.") # Informa ao usuário como sair do jogo.

while jogando: # Inicia um loop 'while' que continua enquanto a variável 'jogando' for True.
    forma_correta = random.choice(list(formas.keys())) # Escolhe aleatoriamente uma forma geométrica (chave do dicionário) para ser a resposta correta da rodada.
    # Seleciona aleatoriamente até 2 dicas para a forma_correta. 'k' é o número de dicas, limitado pelo número total de dicas disponíveis para a forma.
    dicas = random.sample(formas[forma_correta], k=min(len(formas[forma_correta]), 2))

    print("\n--- Nova Rodada ---") # Imprime um separador para indicar o início de uma nova rodada.
    for i, dica in enumerate(dicas): # Itera sobre as dicas selecionadas, atribuindo um índice 'i' e a 'dica' em si.
        print(f"Dica {i+1}: {dica}") # Exibe cada dica para o jogador, numerando-as.

    resposta = input("Qual é a forma? ").lower().strip() # Solicita a resposta do jogador, converte para minúsculas e remove espaços extras.

    if resposta == "sair": # Verifica se o jogador digitou 'sair'.
        jogando = False # Se sim, define 'jogando' como False para encerrar o loop e o jogo.
        print(f"Obrigado por jogar! Sua pontuação final foi: {pontuacao}") # Exibe a mensagem de despedida e a pontuação final.
    elif resposta == forma_correta: # Se a resposta não foi 'sair', verifica se a resposta do jogador está correta.
        print("Correto! Muito bem!") # Se sim, informa que a resposta está correta.
        pontuacao += 1 # Aumenta a pontuação do jogador em 1.
    else: # Se a resposta não foi 'sair' e está incorreta.
        print(f"Errado. A forma correta era: {forma_correta.capitalize()}") # Informa que a resposta está errada e revela a forma correta.

    print(f"Pontuação atual: {pontuacao}") # Exibe a pontuação atual do jogador ao final de cada rodada.
```

API de Consulta de CEP:

```python
import requests  # Importa a biblioteca 'requests' para fazer requisições HTTP a APIs externas.

def consultar_cep(cep):
  """
  Consulta o ViaCEP para obter informações de endereço a partir de um CEP.

  Args:
    cep (str): O número do CEP a ser consultado (apenas dígitos).

  Returns:
    dict: Um dicionário com os dados do endereço, ou None se houver um erro.
  """
  url = f"https://viacep.com.br/ws/{cep}/json/"  # Constrói a URL da API do ViaCEP com o CEP fornecido.
  try:
    response = requests.get(url)  # Faz uma requisição GET para a URL da API.
    response.raise_for_status()  # Verifica se a requisição foi bem-sucedida (status 2xx); caso contrário, lança uma exceção HTTPError.
    data = response.json()  # Converte a resposta JSON em um dicionário Python.
    if 'erro' in data:  # O ViaCEP retorna {'erro': true} se o CEP não for encontrado.
      print(f"Erro: CEP {cep} não encontrado.")  # Exibe uma mensagem de erro se o CEP não for encontrado.
      return None  # Retorna None indicando que não foi possível encontrar o endereço.
    return data  # Retorna o dicionário com os dados do endereço.
  except requests.exceptions.RequestException as e:  # Captura qualquer exceção que possa ocorrer durante a requisição (ex: problemas de conexão).
    print(f"Erro ao conectar ao ViaCEP: {e}")  # Exibe uma mensagem de erro de conexão.
    return None  # Retorna None indicando falha na conexão.

# --- Exemplo de Uso da Função de Consulta de CEP --- #

cep_digitado = input("Digite o CEP para consulta (apenas números, ex: 01001000): ")  # Solicita ao usuário que digite um CEP.

if cep_digitado.isdigit() and len(cep_digitado) == 8:  # Verifica se o CEP digitado contém apenas 8 dígitos numéricos.
  endereco = consultar_cep(cep_digitado)  # Chama a função 'consultar_cep' com o CEP fornecido.
  if endereco:  # Se a função retornar um endereço (ou seja, não for None).
    print("\n--- Endereço Encontrado ---")  # Imprime um cabeçalho para o resultado.
    for chave, valor in endereco.items():  # Itera sobre os pares chave-valor do dicionário de endereço.
      # Imprime cada chave e valor formatados, substituindo '_' por espaço e capitalizando a primeira letra da chave.
      print(f"{chave.replace('_', ' ').title()}: {valor}")
  else:  # Se a função 'consultar_cep' retornou None (CEP não encontrado ou erro de conexão).
    print("Não foi possível obter os dados do endereço.")  # Informa que os dados não puderam ser obtidos.
else:  # Se o CEP digitado não for válido (não tiver 8 dígitos ou não for numérico).
  print("CEP inválido. Por favor, digite 8 dígitos numéricos.")  # Informa ao usuário que o CEP é inválido.
```
