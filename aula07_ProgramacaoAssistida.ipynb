```python
# Define uma lista de números inteiros
num = [1,2,3]

# Inicializa a variável 'soma' para acumular o total
soma = 0

# Inicializa o contador 'i' para iterar sobre a lista
i = 0

# Loop 'while' que continua enquanto 'i' for menor que o comprimento da lista 'num'
while i < len(num):
  # Adiciona o elemento atual da lista (num[i]) à variável 'soma'
  soma += num[i]
  # Incrementa o contador 'i' para passar para o próximo elemento
  i += 1

# Imprime o valor final da soma de todos os elementos da lista
print(soma)
```
